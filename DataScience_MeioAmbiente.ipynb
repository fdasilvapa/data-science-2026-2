```python
# GERADO PELA IA PARA PEGAR COMO EXEMPLO - ajustado para funcionar

import pandas as pd
import requests

url = "https://air-quality-api.open-meteo.com/v1/air-quality?latitude=-16.3267&longitude=-48.9522&hourly=pm10,pm2_5,dust,uv_index,carbon_dioxide,carbon_monoxide,uv_index_clear_sky,ozone,sulphur_dioxide&timezone=America%2FSao_Paulo&domains=cams_global&start_date=2025-01-01&end_date=2025-12-31"

# 2. Puxar os dados da API
response = requests.get(url)
data = response.json()

# 3. Converter para DataFrame do Pandas
df = pd.DataFrame(data["hourly"])

# 4. Formatar a coluna de data/hora para o padrão brasileiro (DD/MM/AAAA HH:MM)
df["time"] = pd.to_datetime(df["time"])


# Configura o pandas para mostrar todas as colunas claramente
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 1000)

print(f"Dados carregados para Anápolis. Total de linhas: {df.shape[0]}\n")

# Printa apenas as 50 primeiras linhas
print(df.head(50))
```

```python
# Uso de head(), info() e describe()

print("=== PRIMEIRAS 5 LINHAS DOS DADOS ===")
display(df.head())

print("\n=== INFORMAÇÕES DO DATAFRAME ===")
df.info()

print("\n=== ESTATÍSTICAS DESCRITIVAS ===")
display(df.describe())

print("\n=== DIMENSÕES DO DATAFRAME ===")
print(f"Total de linhas: {df.shape[0]}")
print(f"Total de colunas: {df.shape[1]}")
```

```python
# Identificação de problemas nos dados, com números (ex.: "12 linhas duplicadas", "8% de nulos")

total_linhas = len(df)
print(f"Total de registros analisados: {total_linhas} horas (365 dias x 24h)\n")

# ------------------------------------------------------------------------------
# 1. A integridade básica para a verificação dos dados se possuem nulos, duplicidades e possiveis continuidades temporais.
# ------------------------------------------------------------------------------
# Modelos de reanálise (CAMS) costumam preencher a grade, mas precisamos provar a ausência de falhas
nulos = df.isnull().sum()
total_nulos = nulos.sum()
print(f"1. Valores Nulos: {total_nulos} células vazias (0.00% em todas as colunas).")

# Checa se a API enviou linhas repetidas ou se há dois registros para o mesmo horário
dup_linhas = df.duplicated().sum()
dup_tempo = df["time"].duplicated().sum()
print(f"2. Duplicidades: {dup_linhas} linhas duplicadas e {dup_tempo} colisões de horário.")

# Verifica se a série temporal é estritamente horária (sem saltos de horas faltantes)
gaps_horario = (df["time"].diff()[1:] != pd.Timedelta(hours=1)).sum()
print(f"3. Continuidade Temporal: {gaps_horario} buracos na sequência de 8.760 horas.")

# ------------------------------------------------------------------------------
# 2. Validação física refente as condições presente no ar, leis da física e do domínio ambiental
# ------------------------------------------------------------------------------
print("\n--- VALIDAÇÃO DE REGRAS FÍSICAS ---")

# Concentrações químicas e grandezas físicas no ar não podem assumir valores negativos (< 0)
colunas_num = df.select_dtypes(include=["float64", "int64"]).columns
total_negativos = sum((df[col] < 0).sum() for col in colunas_num)
print(f"4. Valores Negativos: {total_negativos} valores inválidos detectados.")

# O PM2.5 (partículas finas) é fração do PM10 (partículas inaláveis). PM2.5 > PM10 é fisicamente impossível
violacao_pm = (df["pm2_5"] > df["pm10"]).sum()
pm_iguais = (df["pm2_5"] == df["pm10"]).sum()
print(f"5. Regra PM2.5 <= PM10: {violacao_pm} violações físicas encontradas.")
print(f"   -> Alerta: Em {pm_iguais} horas ({(pm_iguais/total_linhas)*100:.2f}%), PM2.5 == PM10 (ausência de poeira grossa no modelo).")

# À noite não há incidência solar direta; uv_index > 0 entre 20h e 04h indicaria erro de sensor/algoritmo
noite = (df["time"].dt.hour >= 20) | (df["time"].dt.hour <= 4)
uv_noturno = (df.loc[noite, "uv_index"] > 0).sum()
print(f"6. Radiação Noturna: {uv_noturno} leituras anômalas de UV > 0 durante a noite.")

# O índice UV medido não pode superar o potencial máximo teórico de céu sem nuvens
uv_excedente = (df["uv_index"] > df["uv_index_clear_sky"]).sum()
print(f"7. Coerência de Radiação: {uv_excedente} leituras com UV real superando o UV céu limpo.")

# ------------------------------------------------------------------------------
# 3. Detecção estatisticas de outliers: utilizando o IQR (Q3 + 1.5 * IQR)
# ------------------------------------------------------------------------------
print("\n--- OUTLIERS E PICOS EXTREMOS (CRITÉRIO IQR) ---")

outliers = []
for col in colunas_num:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    limite_superior = q3 + 1.5 * iqr
    
    # Conta registros que ultrapassam a cerca estatística superior
    qtd = (df[col] > limite_superior).sum()
    pct = (qtd / total_linhas) * 100
    
    outliers.append({
        "Poluente/Variável": col,
        "Outliers (Qtd)": qtd,
        "Outliers (%)": f"{pct:.2f}%",
        "Limite Superior": round(limite_superior, 2),
        "Pico Máximo": round(df[col].max(), 2)
    })

tabela_outliers = pd.DataFrame(outliers)
print(tabela_outliers.to_string(index=False))
```

```python
# Aplicação da limpeza do pilar do grupo, mostrando antes e depois

df_antes = df.copy()
df_limpo = df.copy()
colunas_num = df_limpo.select_dtypes(include=["float64", "int64"]).columns

# ------------------------------------------------------------------------------
# 1. Integridade básica: duplicatas, nulos e continuidade temporal
#    (o bloco anterior não encontrou problemas, mas a limpeza é aplicada de forma defensiva)
# ------------------------------------------------------------------------------
df_limpo = df_limpo.drop_duplicates()
df_limpo = df_limpo.drop_duplicates(subset="time", keep="first")
df_limpo = df_limpo.sort_values("time").reset_index(drop=True)
# Nulos: interpolação linear no tempo (série contínua horária)
df_limpo[colunas_num] = df_limpo[colunas_num].interpolate(method="linear", limit_direction="both")

# ------------------------------------------------------------------------------
# 2. Regras físicas: negativos, PM2.5 > PM10, UV noturno e UV > céu limpo
# ------------------------------------------------------------------------------
df_limpo[colunas_num] = df_limpo[colunas_num].clip(lower=0)
df_limpo["pm2_5"] = df_limpo["pm2_5"].clip(upper=df_limpo["pm10"])
noite = (df_limpo["time"].dt.hour >= 20) | (df_limpo["time"].dt.hour <= 4)
df_limpo.loc[noite, "uv_index"] = 0
df_limpo["uv_index"] = df_limpo["uv_index"].clip(upper=df_limpo["uv_index_clear_sky"])

# ------------------------------------------------------------------------------
# 3. Outliers (IQR): winsorização, limitando ao teto Q3 + 1.5*IQR em vez de remover linhas.
#    Remover apagaria horas inteiras da série; além disso, picos de poluição
#    (ex.: queimadas) podem ser reais, então apenas suavizamos o extremo.
#    Os limites são calculados sobre os dados ANTES da limpeza.
# ------------------------------------------------------------------------------
#    Exceções: 'dust' (Q3 = 0, o teto zeraria a poeira) e as colunas de UV (zero à noite,
#    então o IQR marca o UV normal do meio-dia como outlier). Esses "outliers" são falsos positivos.
cols_outlier = [c for c in colunas_num if c not in ("dust", "uv_index", "uv_index_clear_sky")]
for col in cols_outlier:
    q1, q3 = df_antes[col].quantile([0.25, 0.75])
    teto = q3 + 1.5 * (q3 - q1)
    df_limpo[col] = df_limpo[col].clip(upper=teto)

# ------------------------------------------------------------------------------
# 4. PM2.5 == PM10: característica do modelo CAMS (sem poeira grossa), não erro de dado.
#    Os valores NÃO são alterados; as linhas são apenas sinalizadas em uma nova coluna.
# ------------------------------------------------------------------------------
df_limpo["pm_igual"] = (df_antes["pm2_5"] == df_antes["pm10"]).values  # calculado nos dados originais
qtd_igual = int(df_limpo["pm_igual"].sum())
print(f"PM2.5 == PM10 sinalizado em {qtd_igual} horas ({qtd_igual / len(df_limpo) * 100:.2f}%), valores mantidos.\n")

# ------------------------------------------------------------------------------
# ANTES x DEPOIS
# ------------------------------------------------------------------------------
def resumo(d, ref):
    noite_d = (d["time"].dt.hour >= 20) | (d["time"].dt.hour <= 4)
    r = {
        "Total de linhas": len(d),
        "Valores nulos": int(d.isnull().sum().sum()),
        "Linhas duplicadas": int(d.duplicated().sum()),
        "Horários duplicados": int(d["time"].duplicated().sum()),
        "Buracos temporais": int((d["time"].diff()[1:] != pd.Timedelta(hours=1)).sum()),
        "Valores negativos": int((d[colunas_num] < 0).sum().sum()),
        "Violações PM2.5 > PM10": int((d["pm2_5"] > d["pm10"]).sum()),
        "UV noturno > 0": int((d.loc[noite_d, "uv_index"] > 0).sum()),
        "UV > UV céu limpo": int((d["uv_index"] > d["uv_index_clear_sky"]).sum()),
    }
    for col in colunas_num:
        q1, q3 = ref[col].quantile([0.25, 0.75])
        r[f"Outliers {col}"] = int((d[col] > q3 + 1.5 * (q3 - q1)).sum())
    return pd.Series(r)

comparacao = pd.DataFrame({"Antes": resumo(df_antes, df_antes), "Depois": resumo(df_limpo, df_antes)})
comparacao["Corrigidos"] = comparacao["Antes"] - comparacao["Depois"]
display(comparacao)

print("\n=== PICO MÁXIMO POR VARIÁVEL (antes x depois) ===")
display(pd.DataFrame({"Máx. antes": df_antes[colunas_num].max().round(2),
                      "Máx. depois": df_limpo[colunas_num].max().round(2),
                      "Média antes": df_antes[colunas_num].mean().round(2),
                      "Média depois": df_limpo[colunas_num].mean().round(2)}))

# ------------------------------------------------------------------------------
# BOXPLOTS ANTES x DEPOIS (variáveis em que os outliers foram tratados)
# ------------------------------------------------------------------------------
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 3, figsize=(14, 7))
for ax, col in zip(axes.flat, ["pm10", "pm2_5", "carbon_monoxide", "ozone", "sulphur_dioxide", "carbon_dioxide"]):
    ax.boxplot([df_antes[col], df_limpo[col]])
    ax.set_xticks([1, 2], ["Antes", "Depois"])
    ax.set_title(col)
fig.suptitle("Distribuição antes e depois do tratamento de outliers")
plt.tight_layout()
plt.show()
```
