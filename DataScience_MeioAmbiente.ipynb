```python
# GERADO PELA IA PARA PEGAR COMO EXEMPLO - ajustado para funcionar

import pandas as pd
import requests

url = "https://air-quality-api.open-meteo.com/v1/air-quality?latitude=-16.3267&longitude=-48.9522&hourly=pm10,pm2_5,dust,uv_index,carbon_dioxide,carbon_monoxide,uv_index_clear_sky,ozone,sulphur_dioxide&timezone=America%2FSao_Paulo&domains=cams_global&start_date=2025-01-01&end_date=2025-12-31"

# 2. Puxar os dados da API
response = requests.get(url)
data = response.json()

# 3. Converter para DataFrame do Pandas
df = pd.DataFrame(data["hourly"])

# 4. Formatar a coluna de data/hora para o padrão brasileiro (DD/MM/AAAA HH:MM)
df["time"] = pd.to_datetime(df["time"])


# Configura o pandas para mostrar todas as colunas claramente
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 1000)

print(f"Dados carregados para Anápolis. Total de linhas: {df.shape[0]}\n")

# Printa apenas as 50 primeiras linhas
print(df.head(50))
```

```python
# Uso de head(), info() e describe()

print("=== PRIMEIRAS 5 LINHAS DOS DADOS ===")
display(df.head())

print("\n=== INFORMAÇÕES DO DATAFRAME ===")
df.info()

print("\n=== ESTATÍSTICAS DESCRITIVAS ===")
display(df.describe())

print("\n=== DIMENSÕES DO DATAFRAME ===")
print(f"Total de linhas: {df.shape[0]}")
print(f"Total de colunas: {df.shape[1]}")
```
